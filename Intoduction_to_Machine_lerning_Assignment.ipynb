1. What is a parameter?
- A parameter is a value that a machine learning model learns from the data during training (like weights in linear regression).
It directly controls how the model makes predictions.
Changing parameters changes the model’s behavior.

2. What is correlation?

Negative correlation means when one variable increases, the other decreases (e.g., price ↑, demand ↓).

3. Define Machine Learning. What are the main components in Machine Learning?
- Machine Learning is a method where computers learn patterns from data and make predictions without being explicitly programmed.

Main components:

4. How does loss value help in determining whether the model is good or not?
- The loss value shows how far the model’s predictions are from the actual values.
A lower loss means the model is performing well, while a high loss means poor accuracy.
During training, the goal is to minimize the loss.

5. What are continuous and categorical variables?

Categorical variables represent groups or labels, not numbers (e.g., gender, color, city).

15. What is causation? Explain difference between correlation and causation with an example.
- Causation means one variable directly causes a change in another.

Difference:

Correlation = variables move together (but not necessarily cause each other).

Causation = one variable causes the other.

Example: Ice cream sales ↑ and drowning ↑ → correlated, but ice cream doesn’t cause drowning; hot weather causes both.

16. What is an Optimizer? What are different types of optimizers? Explain each with an example.
-An optimizer is an algorithm that updates model parameters to minimize the loss during training.

Common types:

Gradient Descent (GD): Updates using the full dataset. Example: linear regression.

Stochastic Gradient Descent (SGD): Updates per sample, faster for large data.

Mini-batch GD: Updates using small batches, balances speed & stability.

Adam: Combines momentum & adaptive learning rates, widely used in deep learning.
